```python
import pandas as pd
import seaborn as sns

sns.get_dataset_names()
```

```python
anagrams = sns.load_dataset(name="anagrams", data_home="./")
with pd.ExcelWriter(path="./seaborn-dataset.xlsx", mode="w") as writer:
    anagrams.to_excel(excel_writer=writer, sheet_name="anagrams")
```

```python
for name in sns.get_dataset_names():
    df = sns.load_dataset(name=name, data_home="./")
    with pd.ExcelWriter(path="./seaborn-dataset.xlsx", mode="a", if_sheet_exists="overlay") as writer:
        df.to_excel(excel_writer=writer, sheet_name=name)
        print(f"{name}\t: successfully exported!")
```
